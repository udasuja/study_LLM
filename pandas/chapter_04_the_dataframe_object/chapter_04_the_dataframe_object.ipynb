## 4.1 Overview of a DataFrame

DataFrame은 행과 열이 있는 2차원 data table이다.

Series와 마찬가지로 index label과 index 위치를 각 행에 할당한다. 또한 각 열에 label과 위치를 할당한다.

DataFrame은 공통 index(=행 번호)를 갖는 Series object의 collection라고 생각하면 쉽게 이해할 수 있다.

```python
import pandas as pd
import numpy as np
```

### 4.1.1 Creating DataFrame from Dictionary

```python
#dictionary의 key값은 DataFrame의 각 column 속성의 이름이 되고,
#value는 DataFrame의 값이 된다.
city_data = {
    "City": ["New York City", "Paris", "Barcelona", "Rome"],
    "Country": ["United States", "France", "Spain", "Italy"],
    "Population": pd.Series([8600000, 2141000, 5515000, 2873000])
}

cities = pd.DataFrame(city_data)
cities
```

```python
# The two lines below are equivalent
#column headers을 index label로 바꾼다.
cities.transpose()
cities.T
```

### 4.1.2 Creating a DataFrame from a NumPy ndarray

```python
random_data = np.random.randint(1, 101, [3, 5])
random_data
```

```python
#ndarray을 값으로 갖는 DataFrame을 만들 수도 있다.
pd.DataFrame(data = random_data)
```

```python
#index 매개변수에 list, tuple, ndarray와 같은 반복 가능한 object를 넘길 수 있으며
#이 매개변수를 통해 원하는 행 label을 설정할 수 있다.
#주의할 점은 label로 할 object의 길이는 DataFrame의 행 개수와 동일해야 한다.

row_labels = ["Morning", "Afternoon", "Evening"]
temperatures = pd.DataFrame(
    data = random_data, index = row_labels
)
temperatures
```

```python
#columns 매개변수를 사용하여 열 이름을 설정할 수 있다.

row_labels = ["Morning", "Afternoon", "Evening"]
column_lables = (
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday"
)

pd.DataFrame(
    data = random_data,
    index = row_labels,
    columns = column_lables,
)
```

```python
#pandas는 중복된 행과 열 index를 허용한다.

row_labels = ["Morning", "Afternoon", "Morning"]
column_lables = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Tuesday",
    "Friday"
]

pd.DataFrame(
    data = random_data,
    index = row_labels,
    columns = column_lables
)
```

## 4.3 Sorting a DataFrame

### 4.3.1 Sorting by Single Column

```python
# The two lines below are equivalent
#Name column을 알바벳 기준으로 오름차순으로 정렬하겠다.
nba.sort_values("Name")
nba.sort_values(by = ["Name"])
```

```python
nba.sort_values("Name", ascending = False).head()
```

```python
nba.sort_values("Birthday", ascending = False).head()
```

### 4.3.2 Sorting by Multiple Columns

```python
#by매개변수에 list를 전달하여 여러 column을 정렬할 수 있다.
#Team을 기준으로 정렬한 다음 이름을 기준으로 정렬한다.
nba.sort_values(by = ["Team", "Name"])
```

```python
nba.sort_values(by=["Team","Birthday"],ascending=False)
```

```python
nba.sort_values(["Team", "Name"], ascending = False)
```

```python
#Team은 오름차순으로 정렬하고, Salary는 내림차순으로 정렬하기 위해, ascending에 bool list를 전달하였다.
nba.sort_values(
    by = ["Team", "Salary"], ascending = [True, False]
)
```

```python
nba = nba.sort_values(
    by = ["Team", "Salary"],
    ascending = [True, False]
)
```
