## 3.1 Importing a Data Set with the read_csv Function

```python
import pandas as pd
```

pandas library에 있는 CSV파일을 가져오는 함수인 read_csv()의 출력값은 항상 2차원 DataFrame이다.


```python
# The two lines below are equivalent
pd.read_csv(filepath_or_buffer = "pokemon.csv")
pd.read_csv("pokemon.csv")
```

```python
#index_col라는 매개변수를 사용하여 DataFrame의 여러 column중 하나를 index label로 설정할 수 있다.
poket=pd.read_csv("pokemon.csv", index_col = "Pokemon")
poket
```

```python
#column의 개수가 1개 밖에 없어도 Series타입이 아닌 DataFrame타입이다.
type(poket)
```

```python
#1개의 column을 갖는 DataFrame을 Series로 변환하고 싶으면, DataFrame 클래스에 있는 squeeze() method을 사용하면 된다. 
pokemon=pd.read_csv("pokemon.csv", index_col = "Pokemon")
print(type(pokemon.squeeze()))
pokemon=pokemon.squeeze()
pokemon
#위 Series의 index label은 pokemon의 이름이고 value는 pokemon의 유형이다.
```

```python
pd.read_csv("google_stocks.csv").head()
```

```python
#parse_dates매개변수는 pandas가 날짜/시간으로 변환해야 하는 텍스트 값의 열을 나타내는 문자열 리스트를 인수로 받는다.
#이때는 "Date"라는 column의 이름이 담긴 문자열 리스트가 인수로 전달되었다.
pd.read_csv("google_stocks.csv", parse_dates = ["Date"]).head()
```

```python
#Date라는 column을 문자열이 아닌 다른 데이터 유형으로 저장하고, 이를 index label로 사용하는 Series을 구성한다.
google=pd.read_csv(
    "google_stocks.csv",
    parse_dates = ["Date"],
    index_col = "Date"
)
google=google.squeeze()
google
```

```python
pd.read_csv("revolutionary_war.csv").tail() 
```

```python
pd.read_csv(
    "revolutionary_war.csv",
    index_col = "Start Date",
    parse_dates = ["Start Date"],
).tail()
```

```python
#Start Date column의 문자열을 날짜/시간 값으로 변환하고, 이를 index label로 활용한다.
#이때의 DataFrame의 column의 개수는 3개인데, Series로 변환하고 싶다면 2개의 column만을 가져와야 한다.
#하나는 index label용이고, 다른 하나는 value용이다.
#usecols라는 매개변수로 CSV파일에서 가져와야 하는 column의 목록을 (문자열 리스트 형태로) 인자로 받는다.
battles = pd.read_csv(
    "revolutionary_war.csv",
    index_col = "Start Date",
    parse_dates = ["Start Date"],
    usecols = ["State", "Start Date"]  
).squeeze()
battles
```

## 3.2 Sorting a Series

### 3.2.1 Sorting by Values with the sort_values Method

```python
#value을 기준으로 오름차순으로 정렬된 새 Series을 반환한다.
google.sort_values()
```

```python
pokemon.sort_values()
```

```python
#pandas는 대문자를 소문자보다 먼저 정렬한다.
pd.Series(data = ["Adam", "adam", "Ben"]).sort_values()
```

```python
#내림차순으로 정렬하고 싶다면 ascending에 False을 전달하면 도니다.
google.sort_values(ascending = False)
```

```python
pokemon.sort_values(ascending = False)
```

```python
# The two lines below are equivalent
battles.sort_values()
#na_position매개변수는 반환된 Series에서 NaN값을 어디에 배치할지를 결정하며, 기본적으로 "last"를 갖는다.
#즉 정렬된 Series는 기본적으로 마지막에 NaN값을 배치한다.
battles.sort_values(na_position = "last")
```

## 3.3 Overwriting a Series with the inplace Parameter

### inplace 매개변수 사용을 권장하지 않는다. 

```python
battles.head(3)
```

```python
battles.sort_values().head(3)
```

```python
battles.head(3)
```

```python
battles.head(3)
```

```python
type(battles)
```

```python
#inplace에 True을 전달하면 method가 호출되는 object를 수정하게 된다.
#이때는 battles이 오름차순으로 정렬된 형태로 수정된다.
battles.sort_values(inplace = True)
```

battles이라는 object가 독립적인 Series가 아닌 DataFrame에서 파생된 Series이기 때문에 다음과 같이 이러한 파생관계를 지워야만 위 코드가 오류없이 실행시킬 수 있다.

```python
battles=battles.copy()
battles.sort_values(inplace=True)
```

```python
battles.head(3)
```

inplace 매개변수를 사용하더라도 pandas는 method을 호출할 때마다 object의 복사본을 만든다.

inplace 매개변수를 True로 설정했을 경우 기존 변수에 새 object을 할당하는 것이므로 당연히 battles.sort_values()의 결과가 object로 메모리에 저장이 되고, 이 object가 battle에 덮어씌어 지는 것이다.


그렇기에 다음 두 코드는 기술적으로 동일하다.

```python
battles.sort_values(inplace=True)
battles=battles.sort_values()  #권장
```

## 3.4 Counting Values with the value_counts Method

```python
pokemon.head()
```

포켓몬의 가장 흔한 유형이 뭔지 알기 위해 다음과 같이 값을 버킷으로 그룹화하고, 각 버킷의 요소 개수를 계산한다.

```python
#각 Series 값의 발생 횟수를 계산하는 value_counts() method사용
#반환되는 Series의 index label은 pokemon Series의 값이고, 값은 pokemon Series의 값의 개수이다.
pokemon.value_counts()
```

```python
len(pokemon.value_counts())
```

```python
pokemon.nunique()
```

```python
pokemon.value_counts(ascending = True)
```

```python
#normalize 매개변수를 True로 설정하여 각 고유값의 빈도를 계산할 수 있다.
pokemon.value_counts(normalize = True).head()
```

```python
#각 포켓몬 유형이 전체에 기여하는 비율을 얻기 위해서는 100을 곱하면된다.
pokemon.value_counts(normalize = True).head() * 100
```

```python
#round()method를 사용하여 Series의 값을 반올림할 수 있다.
#decimals 매개변수는 소수점 뒤에 남겨둘 자릿수를 결정한다.
(pokemon.value_counts(normalize = True) * 100).round(2)
```

```python
#google Series의 값이 숫자값일 때 각 숫자값의 빈도수를 값으로 갖는 Series를 반환한다.
google.value_counts().head()
```

```python
google.max()
```

```python
#주가를 0에서 시작하여 최대 1,400까지 200개의 버킷으로 그룹화하겠다.
buckets = [0, 200, 400, 600, 800, 1000, 1200, 1400]
google.value_counts(bins = buckets)
```

```python
#index label을 기준으로 오름차순으로 정렬
google.value_counts(bins = buckets).sort_index()
```

```python
#sort매개변수를 False로 설정하여, 값을 내림차순으로 정렬하는 것을 막을 수 있다.
google.value_counts(bins = buckets, sort = False)
```

```python
#bins매개변수에 정수 인수가 전달될 경우, 이 정수 값으로 구간이 나눠진다.
#이때 pandas는 Series의 최댓값과 최솟값의 차이를 자동으로 계산하고 범위를 지정된 수의 빈값으로 나눈다.
#즉 단순 0부터 시작하는 것이 아닌 최솟값부터 구간이 시작된다.
google.value_counts(bins = 6, sort = False)
```

## 3.5 Invoking a Function on Every Series Value with the apply Method

파이썬에서는 함수도 다른 데이터 유형과 마찬가지로 유효한 자료구조이다.

즉 숫자로 할 수 있는 모든 것을 함수로 할 수 있다.

        함수를 리스트에 저장할 수 있다.
        dictionary key에 대한 값으로 함수를 할당할 수 있다.
        함수를 다른 함수에 인수로 전달할 수 있다.
        다른 함수에서 함수를 반환할 수 있다.

```python
funcs = [len, max, min]
```

```python
funcs = [len, max, min]

#순서대로 google Series의 행의 개수, 최댓값, 최소값을 출력하게 된다.
for current_func in funcs:
    print(current_func(google))
```

```python
round(99.2)
```

```python
round(99.49)
```

```python
round(99.5)
```

```python
# The two lines below are equivalent
#반올림 함수 round()를 Series의 모든 값에 적용할 수 있다.
#apply() method는 각 Series값에 대해 한 번씩 함수를 호출하고 함수 호출의 반환 값으로 구성된 새 Series를 반환한다.
google.apply(func = round)
google.apply(round)
```

```python
def single_or_multi(pokemon_type):
    if "/" in pokemon_type:
        return "Multi"
    
    return "Single"
```

```python
pokemon.head(4)
```

```python
#사용자 정의 함수를 Series의 각 값에 적용하여, 포켓몬 타입이 여러개인지, 하나인지 확인할 수 있다.
pokemon.apply(single_or_multi)
```

```python
pokemon.apply(single_or_multi).value_counts()
```
