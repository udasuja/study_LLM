```python
import pandas as pd
```

```python
#사용할 column과 그 column의 값을 문자열 값에서 날짜/시간에 대한 TimeStamp object로 강제 변환한다.
data=pd.read_csv("revolutionary_war.csv",
                 usecols=["Start Date"],
                 parse_dates=["Start Date"]
).squeeze()
data
```

```python
import datetime as dt
```

```python
#각 TimeStemp object에 대한 요일을 반환하는 함수
def date(date):
    return date.strftime("%A")
```

```python
#Series내 NaT값을 제거, 위 date함수 적용, 값의 빈도수에 대한 Series반환
data.dropna().apply(func=date).value_counts()
```
