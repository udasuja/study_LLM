```python
#2주간의 판매 데이터를 하나의 DataFrame으로 결합하세요. week1 DataFrame에 "Week 1"라는 키를 할당하고 week2 DataFrame에 "Week 2"라는 키를 할당하세요.
pd.concat(objs=[week1,week2],keys=["Week 1","Week 2"])
```

```python
#2주 동안 매주 식당에서 식사한 고객을 찾으세요.
two_week=week1.merge(week2,how="inner",on="Customer ID").drop_duplicates(subset=["Customer ID"])
two_week.head()
```

```python
#2주동안 매주 식당에서 식사를 하고 같은 음식을 주문한 고객을 찾으세요.
week1.merge(week2,how="inner",on=["Customer ID","Food ID"])
```

```python
#1주차에만 방문한 고객과 2주차에만 방문한 고객이 누군지 확인하세요.
week1.merge(week2,how="outer",on="Customer ID",indicator=True).head()
```
