## 10.2 Concatenating Data Sets

두 DataFrame을 결합하는 방법 중 하나는 하나의 DataFrame을 다른 DataFrame의 끝에 추가하여 결합하는 것이다.

```python
#concat()는 두 DataFrame의 행을 하나의 DataFrame으로 결합하는 함수이다.
#objs매개변수는 리스트에 나타나는 순서대로 연결한다.
#이때 두 DataFrame모두 열의 이름이 같기에 DataFrame은 밑으로 추가된다.
pd.concat(objs = [groups1, groups2])
```

pandas는 groups1과 groups2에 동일한 index번호가 있는 것을 문제 삼지 않기에 연결된 DataFrame에 중복된 index label이 생긴다.

```python
groups1.head()
```

```python
groups2.tail()
```

```python
len(groups1)
```

```python
len(groups2)
```

```python
len(groups1) + len(groups2)
```

```python
#ignore_index매개변수에 True를 전달하면 index번호의 중복을 제거한다.
pd.concat(objs = [groups1, groups2], ignore_index = True)
```

```python
#keys매개변수에 전달된 list의 각 문자열을 objs list과 동일한 index 위치에 있는 DataFrame과 연결한다.
#이것을 통해 index도 중복되지 않게 생성하면서 각 행의 원본 Dataframe을 보존할 수 있다.
pd.concat(objs = [groups1, groups2], keys = ["G1", "G2"])
```

```python
#MultiIndex DataFrame에서 loc접근자를 사용하여 해당 행을 추출할 수도 있다.
ex=pd.concat(objs = [groups1, groups2], keys = ["G1", "G2"])
print(type(ex))
ex.loc[("G2",8326)]
```

```python
groups = pd.concat(objs = [groups1, groups2], ignore_index = True)
```

### 10.2.1 Missing Values in Combined DataFrames

두 DataFrame을 결합할 때 pandas는 dataset이 서로 공유하지 않는 행 label과 열 label의 교차점에 NaN을 배치한다.

```python
sports_champions_A = pd.DataFrame(
    data = [
        ["New England Patriots", "Houston Astros"],
        ["Philadelphia Eagles", "Boston Red Sox"]
    ],
    columns = ["Football", "Baseball"],
    index = [2017, 2018]
)

sports_champions_A
```

```python
sports_champions_B = pd.DataFrame(
    data = [
        ["New England Patriots", "St. Louis Blues"],
        ["Kansas City Chiefs", "Tampa Bay Lightning"]
    ],
    columns = ["Football", "Hockey"],
    index = [2019, 2020]
)
sports_champions_B
```

```python
pd.concat(objs = [sports_champions_A, sports_champions_B])
```

```python
sports_champions_C = pd.DataFrame(
    data = [
        ["Pittsburgh Penguins", "Golden State Warriors"],
        ["Washington Capitals", "Golden State Warriors"]
    ],
    columns = ["Hockey", "Basketball"],
    index = [2017, 2018]
)
sports_champions_C
```

```python
pd.concat(objs = [sports_champions_A, sports_champions_C])
```

```python
#concat()의 axis매개변수에 1혹은 "columns"을 전달하면 열 축에 걸쳐 DataFrame을 결합할 수 있다.
pd.concat(
    objs = [sports_champions_A, sports_champions_C], 
    axis = 1
)
pd.concat(
    objs = [sports_champions_A, sports_champions_C], axis = "columns"
)
```

## 10.3 Left Joins

join연산은 논리적 기준을 적용하여 두 DataFrame 사이에 병합할 행 또는 열을 결정하여 병합한다.

left join은 한 DataFrame의 key를 사용하여 다른 DataFrame의 값을 가져온다.

left join은 기본 DataFrame과 관련된 추가 정보를 제공하는 목적으로 다른 DataFrame을 가져올 때 주로 사용한다.

```python
groups.head(3)
```

```python
categories.head(3)
```

```python
#merge()를 사용하여 하나의 DataFrame을 다른 DataFrame에 병합한다.
#method의 첫 번째 매개변수 right은 추가 정보를 제공할 DataFrame을 전달받는다.
#how매개변수에 join유형을 나타내는 문자열을 전달한다(이때는 left를 전달하여 left join임을 명시한다.).
#on매개변수에 일치 조건에 부합하는 열을 전달한다(DataFrame사이에 열 이름이 동일한 경우에만 사용할 수 있다.).
#이때는 groups에서 categories의 category_id값과 일치하는 값을 찾을 때마다 categories의 열을 가져온다.
#만약 categories에서 groups의 category_id를 찾지 못하면 categories에서 가져온 category_name열에 NaN값을 표시한다.
groups.merge(right=categories, how = "left", on = "category_id").head()
```

## 10.4 Inner Joins

inner join은 두 DataFrame에 모두 존재하는 값을 대상으로 DataFrame을 병합한다.

이때 한 DataFrame에만 존재하는 값은 제외된다.

```python
groups.head(3)
```

```python
categories.head(3)
```

```python
#groups와 categories에서 category_id열의 값이 동일한 행을 대상으로 지정한다.
groups.merge(categories, how = "inner", on = "category_id")
```

병합된 DataFrame은 두 DataFrame에서 값이 일치하는 category_id당 하나의 행을 생성한다.

category_id값이 14인 행은 groups에서는 870개나 되지만 categories에서는 1개밖에 없다.

이때 pandas는 groups의 870개 행을 각각 categories에 잇는 하나의 행과 쌍으로 만들어 병합된 DataFrame에는 총 870개의 행이 생성된다.

만약 ID가 14인 categories가 3개 있는 경우 pandas는 총 870*3=2610개의 행을 생성한다.

```python
groups[groups["category_id"] == 14]
```

```python
categories[categories["category_id"] == 14]
```

## 10.5 Outer Joins

outer join은 두 dataset의 모든 record를 병합한다.

```python
groups.head(3)
```

```python
cities.head(3)
```

```python
#두 DataFrame사이에 열의 이름이 다른 경우 on매개변수가 아닌 다른 매개변수를 사용한다.
#left_on에는 merge를 호출하는 DataFrame의 열 이름을 전달하고
#right_on에는 right매개변수에 전달되는 DataFrame의 열 이름을 전달한다.

#pandas가 city_id와 id열 사이에서 일치하는 값을 찾으면 두 DataFrame의 열을 하나의 행으로 병합한다.
#만약 일치하지 않는다면 city_id에 NaN값을 배치한다.
groups.merge(
    right=cities, how = "outer", left_on = "city_id", right_on = "id"
)
```

```python
#값이 어떤 DataFrame에 속하는지 식별하고 싶다면 merge()의 indicator매개변수에 True를 전달하면 된다.
#값이 groups에 속하면 left_only, cities에 속하면 right_only, 둘 다 속하면 both를 저장한다.
groups.merge(
    cities,
    how = "outer",
    left_on = "city_id",
    right_on = "id",
    indicator = True
)
```

```python
outer_join = groups.merge(
    cities,
    how = "outer",
    left_on = "city_id",
    right_on = "id",
    indicator = True
)

#_merge열에서 값이 right_only인 행 즉 cities DataFrame에만 있는 도시 ID를 추출한다.
in_right_only = outer_join["_merge"] == "right_only"

outer_join[in_right_only].head()
```

## 10.6 Merging on Index Labels

```python
cities.head(3)
```

```python
#join하려는 DataFrame이 primary key를 index에 저장하고자 할 때 그 key를 set_index()의 인자로 전달하면 된다.
#이때는 id열이 index label이 된다. 
cities = cities.set_index(keys="id")
```

```python
cities.head(3)
```

```python
groups.head(3)
```

```python
#right_index에 True를 전달하면 cities의 index label과 groups의 city_id열의 값과 일치하는 행을 조인한다.
groups.merge(
    cities,
    how = "left",
    left_on = "city_id",
    right_index = True
)
```
