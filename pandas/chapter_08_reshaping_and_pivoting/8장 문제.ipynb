```python
import pandas as pd
```

```python
cars=pd.read_csv("used_cars.csv")
cars.head()
```

```python
min_wage=pd.read_csv("minimum_wage.csv")
min_wage.head()
```

```python
#cars에서 가격의 합계를 집계하세요. 행 축에서 연료 유형별로 결과를 그룹화하세요.
cars.pivot_table(index="Fuel",values="Price",aggfunc="sum")
```

```python
#cars에서 자동차의 개수를 aggregate하세요. index축에서 제조업체별로 결과를 그룹화하고 열 축에서 변속기 유형을 그룹화하세요. 행과 열 모두에 대한 소계를 표시하세요.
cars.pivot_table(index="Manufacturer",
                 columns="Transmission",
                 values="Price",
                 aggfunc="count",
                 fill_value=0,
                 margins=True
).head()
```

```python
#cars에서 자동차의 평균 가격을 aggregate하세요. index축에 연도 및 연료 유형별로 결과를 그룹화하고 열 축에 변속기 유형을 그룹홯하세요.
report=cars.pivot_table(index=["Fuel","Year"],
                 columns="Transmission",
                 values=["Price"],
                 aggfunc="mean"
)
report
```

```python
#3번 문제의 DataFrame이 주어졌을 때 열 축에서 행 축으로 변속기 유형을 옮기세요.
report.stack()
```

```python
#min_wage를 wide형식에서 narrow형식으로 변환하세요. 즉 8개의 연도 열에 있는 데이터를 하나의 열로 옮기세요.
year_variable=["2010","2011","2012","2013","2014","2015","2016","2017"]
min_wage.melt(id_vars="State",
              value_vars=year_variable,
              var_name="Year",
              value_name="Wage"
)
```
