```python
import pandas as pd
```

```python
#dataset 최적화
netflix=pd.read_csv("netflix.csv",
                    parse_dates=["date_added"])
netflix
```

```python
netflix.info()
```

```python
netflix["type"].value_counts()
```

```python
netflix["type"]=netflix["type"].astype("category")
```

```python
netflix.info()
```

```python
#제목이 Limitless인 콘텐츠 찾기
title=netflix["title"]=="Limitless"
netflix[title]
```

```python
#Robert Rodriguez감독의 영화
direc=netflix["director"]=="Robert Rodriguez"
movie=netflix["type"]=="Movie"
netflix[direc&movie]
```

```python
#2019-07-31 등록되었거나 Robert Altman가 감독인 콘텐츠
date=netflix["date_added"]=="2019-07-31"
direc=netflix["director"]=="Robert Altman"
netflix[date|direc]
```

```python
#감독이 "Orson Welles","Aditya Kripalani" 또는 "Sam Raimi"인 모든 콘텐츠 찾기
direc=["Orson Welles","Aditya Kripalani", "Sam Raimi"]
filtering=netflix["director"].isin(direc)
netflix[filtering]
```

```python
#2019.5.1~2019.6.1 사이에 등록된 모든 콘텐츠 찾기
find=netflix["date_added"].between(
    "2019-05-01",
    "2019-06-01"
)
netflix[find].sort_values("date_added")
```

```python
#director가 NaN값인 행을 제거
direc_nan=netflix["director"].notnull()
netflix[direc_nan]
```

```python
#keep매개변수를 Flase로 설정하면 date_added열에 중복 항목이 있는 모든 행을 제거한다.
#이를 통해 콘텐츠를 단 하나만 등록한 날짜를 식별할 수 있다.
netflix.drop_duplicates(subset="date_added",keep=False)
```
