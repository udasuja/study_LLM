```python
import pandas as pd
```

```python
citi_bike=pd.read_csv("citibike.csv")
citi_bike.head()
```

```python
citi_bike.info()
```

```python
#start_time과 stop_time열을 문자열 대신 날짜/시간 값을 저장하도록 변환하세요.
for a in ["start_time","stop_time"]:
    citi_bike[a]=pd.to_datetime(citi_bike[a])
```

```python
citi_bike.info()
```

```python
#각 요일에 승차한 건 수를 계산하세요. 사람들이 자전거를 가장 많이 타는 요일은 언제인가요? start_time열을 기준으로 계산하세요.
bike_day=citi_bike["start_time"].dt.day_name()
```

```python
bike_day.value_counts().sort_values().head(1)
```

해당 월의 각 주당 승차 건을 구하세요. 이것을 구하려면 start_time열의 각 날짜를 이전 또는 현재 월요일로 변환해야 한다.

한 주가 월요일에서 시작하여 일요일에 끝난다고 가정하세요. 즉, 6월의 첫째주는 6월 1일 월요일부터 6월 7일 일요일까지입니다.

```python
citi_bike["start_time"].dt.dayofweek.head()
```

```python
#각 요일이 월요일과 얼만큼 떨어져 있는지 하루에 1씩 더하여 나타낸다.
days_away_from_monday=citi_bike["start_time"].dt.dayofweek
```

```python
citi_bike["start_time"]-pd.to_timedelta(
    days_away_from_monday,unit="day")
```

```python
dates_rounded_to_monday=citi_bike["start_time"]-pd.to_timedelta(
    days_away_from_monday,unit="day")
```

```python
dates_rounded_to_monday.dt.date.head()
```

```python
#6월 한 달동안 사람들이 자전거를 가장 많이 탄 주는 15일 월부터 21일 일까지이다.
dates_rounded_to_monday.dt.date.value_counts()
```
