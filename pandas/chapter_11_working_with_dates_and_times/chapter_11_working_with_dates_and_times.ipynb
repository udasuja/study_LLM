### 11.1.2 How `pandas` works with datetimes

pandas는 datetime class를 대체하기 위해 Timestamp class를 도입했다.

Timestamp object는 날짜와 더불어 내부적으로 시간도 저장한다.

```python
#Timestamp object와 datatime object생성법은 똑같다.
# The two lines below are equivalent
pd.Timestamp(1991, 4, 12)
pd.Timestamp(year = 1991, month = 4, day = 12)
```

```python
#Timestamp object와 date의 date object가 같은지 확인할 수도 잇다.
#날짜 혹은 시간이 다르면 두 object는 동일하지 않다.
(pd.Timestamp(year = 1991, month = 4, day = 12) 
 == dt.date(year = 1991, month = 4, day = 12))
```

```python
(pd.Timestamp(year = 1991, month = 4, day = 12, minute = 2)
 == dt.datetime(year = 1991, month = 4, day = 12, minute = 2))
```

```python
(pd.Timestamp(year = 1991, month = 4, day = 12, minute = 2) 
 == dt.datetime(year = 1991, month = 4, day = 12, minute = 1))
```

```python
#YYYY-MM-DD형식으로 날짜를 나타내도 된다.
pd.Timestamp("2015-03-31")
```

```python
#-대신 /를 사용해도 된다.
pd.Timestamp("2015/03/31")
```

```python
#MM/DD/YYYY형식의 문자열을 전달해도 된다.
pd.Timestamp("03/31/2015")
```

```python
pd.Timestamp("2021-03-08 08:35:15")
```

```python
pd.Timestamp("2021-03-08 6:13:29 PM")
```

```python
#Timestamp생성자는 python의 기본 date, time 및 datetime object도 허용한다.
pd.Timestamp(dt.datetime(2000, 2, 3, 21, 35, 22))
```

```python
my_time = pd.Timestamp(dt.datetime(2000, 2, 3, 21, 35, 22))
print(my_time.year) 
print(my_time.month)
print(my_time.day)
print(my_time.hour) 
print(my_time.minute)
print(my_time.second)
```

## 11.2 Storing Multiple Timestamps in a DatetimeIndex

index는 pandas 자료구조에 붙여지는 식별자 label collection을 의미한다. 

지금까지는 오름차순 혹은 내림차순으로 정렬된 숫자 값의 시퀀스인 RangeIndex를 사용했다.

```python
#index속성으로 Series의 index에 접근
pd.Series([1, 2, 3]).index
```

pandas는 Index object를 사용해서 문자열 label collection을 저장한다.

```python
#index속성으로 DataFrame의 index에 접근
pd.Series([1, 2, 3], index = ["A", "B", "C"]).index
```

DatatimeIndex는 Timestamp object를 저장하는 index이다.

```python
#Series의 index를 DatetimeIndex로 설정하였다.
timestamps = [
    pd.Timestamp("2020-01-01"),
    pd.Timestamp("2020-02-01"),
    pd.Timestamp("2020-03-01"),
]

pd.Series([1, 2, 3], index = timestamps).index
```

```python
pd.Series([1, 2, 3], index = timestamps)
```

```python
datetimes = [
    dt.datetime(2020, 1, 1),
    dt.datetime(2020, 2, 1),
    dt.datetime(2020, 3, 1),
]

pd.Series([1, 2, 3], index = datetimes).index
```

```python
#다음과 같이 DatetimeIndex자체를 생성할 수도 있다.
#이때 매개변수는 모든 반복 가능한 날짜 collection을 인수로 전달받는다.
#날짜를 문자열, 날자/시간, Timestamp 또는 혼합 데이터 유형으로 전달할 수 있다.
#pandas는 모든 값을 동등하게 Timestmp로 변환하고 index에 저장한다.
string_dates = ["2018/01/02", "2016/04/12", "2009/09/07"]
pd.DatetimeIndex(string_dates)
```

```python
mixed_dates = [
    dt.date(2018, 1, 2),
    "2016/04/12",
    pd.Timestamp(2009, 9, 7)
]

dt_index = pd.DatetimeIndex(mixed_dates)
dt_index
```

```python
#이제 Series의 index에 Timestamp object를 저장하는 index인 DatatimeIndex를 저장한다.
s = pd.Series(data = [100, 200, 300], index = dt_index)
s
```

pandas에서는 날짜/시간 값을 문자열이 아닌 Timestamp로 저장해야만 날짜 및 시간과 관련된 연산을 적용할 수 있다.

```python
s.sort_index()
```

```python
morning = pd.Timestamp("2020-01-01 11:23:22 AM")
evening = pd.Timestamp("2020-01-01 11:23:22 PM")

morning < evening
```

## 11.6 Date Offsets

```python
disney["Date"].tail()
```

```python
#offset object의 MonthEnd object는 각 날짜를 다음 달 말일로 올린다.
#이미 말일인 2020-6-30은 다음달 말일인 2020-07-31로 올린다.
(disney["Date"] + pd.offsets.MonthEnd()).tail()
```

```python
#이때 빼기 기호는 각 날짜를 시간상 뒤로 이동시킨다. 즉 이전 달의 말일로 내린다.
(disney["Date"] - pd.offsets.MonthEnd()).tail()
```

```python
#MonthBegin은 월의 첫 번째 날짜로 올린다.
#+기호와 사용시 각 날짜를 다음달 시작일로 올린다.
(disney["Date"] + pd.offsets.MonthBegin()).tail()
```

```python
#-기호와 사용시 각 날짜를 월의 시작일로 내린다.
#만약 2020-07-01인 경우 2020-06-01로 내린다.
(disney["Date"] - pd.offsets.MonthBegin()).tail()
```

## 11.7 The `timedelta` Object

```python
duration = pd.Timedelta(
    days = 8,
    hours = 7,
    minutes = 6,
    seconds = 5
)

duration
```

```python
#to_timedelta()는 전달된 인수를 Timedelta object로 변환한다.
pd.to_timedelta("3 hours, 5 minutes, 12 seconds")
```

```python
#unit 매개변수는 숫자가 나타내는 시간 단위를 정의한다.
pd.to_timedelta(5, unit = "hour")
```

```python
pd.to_timedelta([5, 10, 15], unit = "day")
```

```python
#한 Timestamp에서 다른 Timestamp를 빼면 자동으로 Timedelta가 반환된다.
pd.Timestamp("1999-02-05") - pd.Timestamp("1998-05-24")
```

```python
deliveries = pd.read_csv("deliveries.csv")
deliveries.head()
```

```python
#to_datetime()를 통해 문자열을 datetime object로 변환한다.
deliveries["order_date"] = pd.to_datetime(
    deliveries["order_date"]
)

deliveries["delivery_date"] = pd.to_datetime(
    deliveries["delivery_date"]
)
```

```python
#반복문을 사용하여 위 과정을 수행하였다.
for column in ["order_date", "delivery_date"]:
    deliveries[column] = pd.to_datetime(deliveries[column])
```

```python
deliveries.head()
```

```python
#delivery_date열에서 order_date열을 빼서 각 배송 기간을 계산할 수도 있다.
#그럼 계산 결과가 timedelta Series로 반환된다.
(deliveries["delivery_date"] - deliveries["order_date"]).head()
```

```python
print(type((deliveries["delivery_date"] - deliveries["order_date"])))
print(type((deliveries["delivery_date"] - deliveries["order_date"])[1]))
```

```python
deliveries["duration"] = (
    deliveries["delivery_date"] - deliveries["order_date"]
)
deliveries.head()
```

```python
deliveries.dtypes
```

```python
(deliveries["delivery_date"] - deliveries["duration"]).head()
```

```python
(deliveries["delivery_date"] + deliveries["duration"]).head()
```

```python
deliveries.sort_values("duration")
```

다음과 같이 Timedelta Series에서도 수학 method를 사용할 수 있다.

```python
deliveries["duration"].max()
```

```python
deliveries["duration"].min()
```

```python
deliveries["duration"].mean()
```
