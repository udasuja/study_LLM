Address열을 Street, City, State, Zip(도로, 시, 주 및 우편번호)이라는 네 가지 값으로 분리하고 기존 Address열을 삭제하라

```python
import pandas as pd
```

```python
customers=pd.read_csv("customers.csv")
customers
```

```python
customers["Address"].head().values
```

```python
customers["Address"].str.split(", ").head()
```

```python
new_dataframe=customers["Address"].str.split(pat=", ",expand=True)
new_dataframe
```

```python
customers[["Street","City","State","Zip"]]=new_dataframe
customers
```

```python
customers=customers.drop(labels=["Address"],axis="columns")
customers
```
