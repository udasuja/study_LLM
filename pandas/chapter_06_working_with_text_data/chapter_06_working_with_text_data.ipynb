## 6.2 String Slicing

```python
inspections["Risk"].head()
```

```python
len(inspections)
```

```python
#Risk Series의 고유한 값을 배열로써 나타내는 unique()로 확인한 결과
#이 Series는 결측값(NaN)과 ALL문자열이라는 예상하지 못한 형식도 포함하고 있는 것을 알 수 있다.
inspections["Risk"].unique()
```

```python
#Risk Series에서 결측값을 갖는 행을 제거한다.
inspections = inspections.dropna(subset = ["Risk"])
```

```python
inspections["Risk"].unique()
```

```python
#replace()는 DataFrame내 to_replace에 대한 값을 value값으로 치환하는 method이다.
inspections["Risk"] = inspections["Risk"].replace(
    to_replace = "All", value = "Risk 4 (Extreme)"
)
```

```python
inspections["Risk"].unique()
```

### 6.3 String Slicing and Character Replacement

StringMethods object의 slice()를 사용하면 index 위치를 기반으로 문자열에서 하위 문자열을 추출할 수 있다.

이 method는 시작 index와 끝 index를 인수로 받는다. 

```python
#Risk 3과 같이 "Risk "뒤 위험도 숫자에 대한 문자를 가져온다.
inspections["Risk"].str.slice(5, 6).head()
```

```python
#python의 list slicing구문으로 동일하게 구현할 수도 있다.
inspections["Risk"].str[5:6].head()
```

```python
#slice()에 단일 값을 전달하면 pandas는 이를 하한값으로 사용하고, 문자열 끝까지 추출한다.
inspections["Risk"].str.slice(8).head()
```

```python
inspections["Risk"].str[8:].head()
```

```python
#문자열 끝을 기준으로 잡아 index 경계를 설정한다.
inspections["Risk"].str.slice(8, -1).head()
```

```python
inspections["Risk"].str[8:-1].head()
```

```python
#replace()를 사용하여 )를 빈 문자열로 치환한다.
inspections["Risk"].str.slice(8).str.replace(")", "").head()
```

## 6.3 Boolean Methods

```python
"Pizza" in "Jets Pizza"
```

```python
"pizza" in "Jets Pizza"
```

```python
#일단 모든 열 값을 소문자로 변환한다. 그리고 소문자 단어 pizza를 찾는다.
#contains()는 각 Series 값에 하위 문자열이 포함되어 있는지 확인한다.
inspections["Name"].str.lower().str.contains("pizza").head()
```

```python
#Name Series 중 Pizza가 포함된 모든 행을 추출한다.
has_pizza = inspections["Name"].str.lower().str.contains("pizza")
inspections[has_pizza]
```

```python
#startswith()는 문자열이 인수로 주어진 문자열로 시작하면 True로 반환하는 method이다.
#이때는 tacos로 시작하는 행을 추출한다.
inspections["Name"].str.lower().str.startswith("tacos").head()
```

```python
starts_with_tacos = (
    inspections["Name"].str.lower().str.startswith("tacos")
)
inspections[starts_with_tacos]
```

```python
#endswith()는 인수로 주어진 문자열이 문자열의 끝에 있는지 확인하는 method이다.
ends_with_tacos = (
    inspections["Name"].str.lower().str.endswith("tacos")
)
inspections[ends_with_tacos]
```

## 6.4 Splitting Strings

```python
customers = pd.read_csv("customers.csv")
customers.head()
```

```python
#str.len()를 사용하여 각 행의 문자열 길이를 반환한다.
customers["Name"].str.len().head()
```

```python
#split()는 인자로 주어진 기호를 사용하여 문자열을 분할하는 method이다.
phone_number = "555-123-4567"
phone_number.split("-")
```

```python
#각 문자열에서 공백을 기준으로 값을 분할한다.
# The two lines below are equivalent
customers["Name"].str.split(pat = " ").head()
customers["Name"].str.split(" ").head()
```

```python
#각 문자열을 공백을 기준으로 값을 분할했을 때 그 결과로 나온 list의 길이를 살펴본다.
customers["Name"].str.split(" ").str.len().head()
```

```python
#분할의 최대 임계값을 1로 설정하여 첫 번째 공백에서 문자열을 분할하고 중지하도록 한다. 
customers["Name"].str.split(pat = " ", n = 1).head()
```

```python
#str.get()을 사용하여 문자열 list내 index 위치를 기반으로 값을 가져올 수 있다.
#다음은 index 0을 대상으로 지정하여 각 list의 첫 번째 요소를 추출한다.
customers["Name"].str.split(pat = " ", n = 1).str.get(0).head()
```

```python
customers["Name"].str.split(pat = " ", n = 1).str.get(1).head()
```

```python
customers["Name"].str.split(pat = " ", n = 1).str.get(-1).head()
```

```python
#expand매개변수에 True를 전달하여 문자열 list를 행으로 갖는 Series가 아닌 DataFrame으로 결과를 반환한다.
customers["Name"].str.split(
    pat = " ", n = 1, expand = True
).head()
```

```python
#index 2를 보면 알 수 있듯이 pandas는 요소가 불충분한 행에는 None값을 배치한다.
customers["Name"].str.split(pat = " ", expand = True).head()
```

```python
#기존 DataFrame에 새로운 2열 DataFrame을 연결한다.
customers[["First Name", "Last Name"]] = customers[
    "Name"
].str.split(pat = " ", n = 1, expand = True)
```

```python
customers
```

```python
#기존에 있던 Name 열을 삭제한다.
customers = customers.drop(labels = "Name", axis = "columns")
```

```python
customers.head()
```
