 Graphical Spreadsheet Applications은 주로 작은 dataset을 처리할 때 사용하면 유용하고, pandas는 아주 큰 dataset을 처리할 때 유용하다.

### 1.3.3 Counting Values in a Series

```python
#영화 제목이 index이고 이에 대응하는 값이 Studio인 Series을 추출하였다.
movies["Studio"]
```

```python
#각 Studio값에 대한 Series내에서의 빈도수를 추출한다.
#Studio가 index label로 사용되고, 빈도 수가 값으로 사용되는 Series을 출력한다.
s_counts=movies["Studio"].value_counts().head(10)
print(type(s_counts))
s_counts
```

### 1.3.4	Filtering a Column by One or More Criteria

```python
#Universal에서 제작된 영화만 추출(DataFrame형식)
movies[movies["Studio"] == "Universal"]
```

```python
released_by_universal = (movies["Studio"] == "Universal")
movies[released_by_universal].head()
```

```python
#여러 기준으로 DataFrame행을 filtering할 수도 있다.
#다음은 universal에서 제작되고, 2015년도에 개봉한 영화를 추출한다.
released_by_universal = movies["Studio"] == "Universal"
released_in_2015 = movies["Year"] == 2015
movies[released_by_universal & released_in_2015]
```

```python
#다음은 universal에서 제작되었거나, 2015년도에 개봉한 영화를 추출한다.
released_by_universal = movies["Studio"] == "Universal"
released_in_2015 = movies["Year"] == 2015
movies[released_by_universal | released_in_2015]
```

```python
before_1975 = movies["Year"] < 1975
movies[before_1975]
```

```python
#1983년에서 1986년 사이에 개봉한 영화를 filtering
mid_80s = movies["Year"].between(1983, 1986)
movies[mid_80s]
```

```python
#DataFrame의 index을 소문자로 바꾸고, index 중에서 dark가 포함된 모든 영화를 추출한다.
has_dark_in_title = movies.index.str.lower().str.contains("dark")
movies[has_dark_in_title]
```

### 1.3.5 Grouping Data

```python
#기호 $과 ,을 모두 빈 텍스트로 바꾼다.
movies["Gross"].str.replace(
    "$", "", regex = False  #$를 ""로 바꾼다는 의미이다.
).str.replace(",", "", regex = False)
```

```python
#밑의 괄호는 별거 아니다.
#movies["Gross"].str.replace("$", "", regex = False).str.replace(",", "", regex = False).astype(float)
#와 같이 쓸 수 있는 것을 보기 편하게 쓰기 위해서 괄호를 활용한 것이다.
(
    movies["Gross"]
    .str.replace("$", "", regex = False)
    .str.replace(",", "", regex = False)
    .astype(float)
)
```

```python
#float64형으로 타입을 변형한 것을 원본에다가 덮어씌운다.
movies["Gross"] = (
    movies["Gross"]
    .str.replace("$", "", regex = False)
    .str.replace(",", "", regex = False)
    .astype(float)
)
```

```python
movies["Gross"].mean()
```

```python
#Studio column의 값을 기반으로 DataFrame의 행을 그룹화한다.
#즉 같은 제작사가 만든 영화를 하나의 그룹으로 만든다는 것이다.
studios = movies.groupby("Studio")
```

groupby()는 테이블을 다시 그룹으로 묶어서 정렬하여 새로운 테이블을 만들지는 않는다.

단지 어떤 Studio에 어떤 행들이 속하는지에 대한 매핑정보만 내부적으로 들고 있는 객체인 DataFrameGroupBy를 만들 뿐이다.

그렇기에 studios.head()로 보는 테이블은 movies로 보는 테이블과 완전히 같다.

```python
studios.head()
```
