## 7.1 The MultiIndex Object

MuliIndex는 관계형db에서 다중 key와 같은 역할을 한다.

```python
import pandas as pd
```

```python
address = ("8809 Flair Square", "Toddside", "IL", "37206")
address
```

```python
addresses = [
    ("8809 Flair Square", "Toddside", "IL", "37206"),
    ("9901 Austin Street", "Toddside", "IL", "37206"),
    ("905 Hogan Quarter", "Franklin", "IL", "37206"),
]
```

```python
#pandas내 MultiIndex class에는 from_tuples라는 class가 있다.
#이는 tuple list로부터 MultiIndex를 instance화할 수 있다.
#다음은 addresses list를 넘겨 from_tuples class method를 호출한다.


# The two lines below are equivalent
pd.MultiIndex.from_tuples(addresses)
pd.MultiIndex.from_tuples(tuples = addresses)
```

위 코드의 결과로 4개의 요소(level)로 구성된 3개의 tuple을 저장하는 MultiIndex를 생성하였다.

예를 들어'8809 Flair Square', '9901 Austin Street', '905 Hogan Quarter'이 같은 level에 속한다.

```python
#from_tuples()의 names 매개변수에 list를 전달하면 각 MultiIndex level에 이름을 할당할 수 있다.
row_index = pd.MultiIndex.from_tuples(
    tuples = addresses,
    names = ["Street", "City", "State", "Zip"]  
)

row_index
```


MultiIndex는 각 label이 여러 값을 보유하는 container이다. level은 label에서 동일한 위치에 있는 값으로 구성된다.


```python
#위 row_index를 사용하기 위해서는 3개의 행이 있는 dataset이 필요하다.
data = [
    ["A", "B+"],
    ["C+", "C"],
    ["D-", "A"],
]

columns = ["Schools", "Cost of Living"]

#DataFrame의 index 매개변수에 MultiIndex를 전달하여 MultiIndex를 DataFrame에 연결할 수 있다.
area_grades = pd.DataFrame(
    data = data, index = row_index, columns = columns
)

area_grades
```

위 DataFrame의 index는 MultiIndex이므로, 각 행은 MultiIndex의 각 요소로 구분된다.

```python
area_grades.columns
```

```python
column_index = pd.MultiIndex.from_tuples([
    ("Culture", "Restaurants"),
    ("Culture", "Museums"),
    ("Services", "Police"),
    ("Services", "Schools"),
])

column_index
```

```python
#위 열 축(column_index)에 대한 MultiIndex를 사용하기 위해서는 4개의 열이 있는 dataset이 필요하다.
data = [
    ["C-", "B+", "B-", "A"],
    ["D+", "C", "A", "C+"],
    ["A-", "A", "D+", "F"], 
]
```

```python
#행과 열 축 모두 MultiIndex를 사용하여 DataFrame을 생성할 수 있다.
pd.DataFrame(
    data = data, index = row_index, columns = column_index
)
```

MultiIndex는 여러 level, 여러 계층을 저장할 수 있는 index 이다. 그리고 각 index label은 여러 구성 요소를 가진다.

```python
#원본 csv파일이라고 하겠다.
neighborhoods = pd.read_csv("neighborhoods.csv")
neighborhoods.head()
```

```python
#가장 왼쪽에 있는 3개의 열이 DataFrame의 index 역할을 한도록 한다.
#이를 위해 그에 해당하는 열의 index를 나타내는 숫자 list를 매개변수 index_col에 전달해야 한다.
#incdex_col에 여러 값이 있는 list가 전달되면 이에 대한 MultiIndex를 자동으로 생성한다.
neighborhoods = pd.read_csv(
    "neighborhoods.csv",
    index_col = [0, 1, 2]
)

neighborhoods.head()
```

```python
#다음으로 DataFrame의 header에 사용할 dataset 행을 pandas에게 알려준다.
#위 원본 csv파일을 봤을 때 0행과 1행을 column header로 설정해야 하는데, 이를 위해 header에 [0,1]를 전달한다.
#그럼 열에 MultiIndex를 할당한다.
neighborhoods = pd.read_csv(
    "neighborhoods.csv",
    index_col = [0, 1, 2],     #0,1,2번째 열이 MultiIndex로써 사용되고
    header = [0, 1]            #0행과 1행이 각각 column header로 사용된다.
)

neighborhoods.head()
```

## 7.3 Sorting a MultiIndex

pandas는 순서가 뒤죽박죽인 컬렉션보다 정렬된 컬렉션에서 훨씬 빠르게 값을 찾을 수 있다.

```python
#MultiIndex DataFrame에서 sort_index()를 호출할 때 pandas는 모든 level을 오름차순으로 정렬하고 밖에서부터 안쪽 방향으로 정렬을 진행한다.
#지금도 보면 State를 먼저 정렬하고, 정렬된 State를 기준으로 City를 정렬 또 이를 기준으로 Street를 정렬한다.
neighborhoods.sort_index()
```

```python
#ascending로 오름차순으로 정렬할지 내림차순으로 정렬할지 결정한다.
neighborhoods.sort_index(ascending = False).head()
```

```python
#각 level의 정렬 순서를 서로 다르게 지정할 수도 있다.
#State는 오름차순, City는 내림차순, Street는 오름차순으로 정렬하도록 설정하였다.
neighborhoods.sort_index(ascending = [True, False, True]).head()
```

```python
#MultiIndex level자체를 정렬할 수도 있다. 이때는 City의 값을 기준으로 행을 정렬하겠다.
#level매개변수에 level의 index위치나 이름을 전달한다.
#이때는 City level만 오름차순으로 정렬된다.

# The two lines below are equivalent
neighborhoods.sort_index(level = 1)
neighborhoods.sort_index(level = "City")
```

```python
#다음은 City와 Street level의 값만 정렬하도록 한다. City정렬하고 이를 기준으로 Street를 정렬하도록 한다.

# The two lines below are equivalent
neighborhoods.sort_index(level = [1, 2]).head()
neighborhoods.sort_index(level = ["City", "Street"]).head()
```

```python
#City는 오름차순으로 Street는 내림차순으로 정렬하도록 한다(2,3번째 행을 보면 어떻게 적용되었는지 알 수 있다.).
neighborhoods.sort_index(
    level = ["City", "Street"], ascending = [True, False]
).head()
```

```python
#정렬할려는 level의 수와 ascending에 전달되는 요소의 개수가 다를 때는 오류가 발생한다.
neighborhoods.sort_index(
    level = ["City", "Street"], ascending = [True,True, False]
).head()
```

```python
#축을 column으로 설정하면 열의 MultiIndex도 정렬할 수 있다.
#다은은 Category level을 먼저 정렬하고 Subcategory level을 두 번째로 정렬한다.

# The two lines below are equivalent
neighborhoods.sort_index(axis = 1).head(3)
neighborhoods.sort_index(axis = "columns").head(3)
```

```python
#Subcategory level 값만을 내림차순으로 정렬한다.
#그럼 Restaurants	Museums	Police	Schools 이 Schools	Restaurants	Police	Museums 로 정렬되는데
#Category그룹이 번갈아 나타나기에 다음과 같이 출력이 된다.

neighborhoods.sort_index(
    axis = 1, level = "Subcategory", ascending = False
).head(3)
```

```python
neighborhoods = neighborhoods.sort_index(ascending = True)
```

## 7.5 Cross Sections

```python
#xs()에 하나의 MultiIndex level에 대한 값을 넘기면 그에 대한 행을 추출할 수 있다.
#찾고자 하는 값을 key매개변수에 전달하고 level에는 index level의 이름이나 숫자 위치를 전달한다.

# The two lines below are equivalent
neighborhoods.xs(key = "Lake Nicole", level = 1)
neighborhoods.xs(key = "Lake Nicole", level = "City")
```

```python
#열의 MultiIndex에 있는 Subcategory level에서 Museums key가 있는 열을 선택한다.

neighborhoods.xs(
    axis = "columns", key = "Museums", level = "Subcategory"
).head()
```

```python
neighborhoods.xs(
    axis="columns",key="Services",level="Category").head()
```

```python
#State값이 AK이고 Street값이 238 Andrew Rue인 행을 찾는다.

# The two lines below are equivalent
neighborhoods.xs(
    key = ("AK", "238 Andrew Rue"), level = ["State", "Street"]
)

neighborhoods.xs(
    key = ("AK", "238 Andrew Rue"), level = [0, 2]
)
```

## 7.6. Manipulating the Index

### 7.6.1 Resetting the Index

```python
neighborhoods.head()
```

```python
#reorder_levels()은 MultiIndex level을 지정된 순서로 정렬한다.
#이때 순서는 order매개변수에 list형태로 전달하면 된다.
new_order = ["City", "State", "Street"]
neighborhoods.reorder_levels(order = new_order).head()
```

```python
#위와 같은 결과를 출력한다.
neighborhoods.reorder_levels(order = [1, 0, 2]).head()
```

```python
#reset_index()는 이전 MultiIndex level을 열로 통합하는 새 DataFrame을 반환한다.

#결과를 보면 각 행은 index label이 아닌 index번호로 구분되는 것을 알 수 있다.
neighborhoods.reset_index().tail()
```

```python
#State, City, Street 3개의 열을 다른 MultiIndex level에 추가할 수도 있다.
#원하는 level의 index 위치 또는 이름을 reset_index()의 col_level매개변수로 전달한다.

# The two lines below are equivalent
neighborhoods.reset_index(col_level = 1).tail()
neighborhoods.reset_index(col_level = "Subcategory").tail() 
```

```python
#State, City, Street의 상위 level Category에 빈 문자열을 할당하게 된다.
#이때 col_fill매개변수에 인수를 전달하면 빈 문자열을 원하는 값으로 바꿀 수 있다.
neighborhoods.reset_index(
    col_fill = "Address", col_level = "Subcategory"
).tail()
```

```python
#level매개변수에 이름을 전달하여 그 이름에 대한 index level만 열로 옮길 수 있다.
neighborhoods.reset_index(level = "Street").tail()
```

```python
neighborhoods.reset_index(level = ["Street", "City"]).tail()
```

```python
#reset_index()의 drop 매개변수 값을 True로 전달하면 pandas는 지정된 level을 열에 추가하는 대신 삭제한다.
neighborhoods.reset_index(level = "Street", drop = True).tail()
```

```python
neighborhoods = neighborhoods.reset_index()
```
