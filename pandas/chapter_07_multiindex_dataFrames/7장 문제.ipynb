```python
import pandas as pd
```

```python
investments=pd.read_csv("investments.csv")
investments
```

```python
investments.nunique()
```

```python
investments.set_index(keys=["Status","State","Funding Rounds"])
```

```python
investments=investments.set_index(keys=["Status","State","Funding Rounds"]).sort_index()
investments
```

```python
#Status가 "Closed"인 모든 행을 추출하세요
investments.loc[("Closed",)]
```

```python
#Status가 Acquired이고 Funding Rounds가 10인 모든 행 추출
investments.xs(key=("Acquired",10),level=("Status","Funding Rounds"))
```

```python
#Status가 Operating이고 Funding Rounds가 6이고 State가 NJ인 모든 행을 추출하세요
investments.xs(key=("Operating","NJ",6),level=("Status","State","Funding Rounds"))
```

```python
#Status가 Closed이고 Funding Rounds가 8인 모든 행을 추출하세요. 이때 Name열만 추출하세요.
investments.reorder_levels(order=["Status","Funding Rounds","State"]).loc[("Closed",8),("Name")]
```

```python
#Status나 Funding Rounds의 값과 상관없이 State가 NJ인 모든 행을 추출하세요.
investments.xs(key=("NJ"),level=("State"))
```

```python
#MultiIndex level을 DataFrame에 열로 다시 통합하세요.
investments=investments.reset_index()
investments
```
